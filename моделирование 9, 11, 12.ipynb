```python
import random
import numpy as np

def game(X, Y, Z):
    a, b, c = X, Y, Z
    rounds = 0
    
    while a > 0 and b > 0 and c > 0:
        # каждый раунд
        roll = random.randint(1, 3)  # 1,2,3 с равной вероятностью
        if roll == 1:
            a += 2
            b -= 1
            c -= 1
        elif roll == 2:
            a -= 1
            b += 2
            c -= 1
        else:  # roll == 3
            a -= 1
            b -= 1
            c += 2
        rounds += 1
    return rounds

def monte_carlo(X, Y, Z, num=100000):
    rounds_2 = 0
    for i in range(num):
        rounds_2 += game(X, Y, Z)
    return rounds_2 / num

def func(X, Y, Z):
    return (X * Y * Z) / (X + Y + Z - 2)

print(monte_carlo(2, 8, 2, 50000))
print(func(2, 8, 2))
```

```python
# 11
def lcg(s, a, c, m):
    return (s * a + c) % m

def main():
    a, c, s = input().split()
    a = int(a)
    c = int(c)
    s = int(s)
    
    m = 2**32
    for i in range(10):
        s = lcg(s, a, c, m)
        print(s)

main()
```

```python
# 12
def lcg_recursive(x0, a, b, m, n, count=0):
    if count >= n:
        return

    x_next = (a * x0 + b) % m
    print(x_next)
    
    lcg_recursive(x_next, a, b, m, n, count + 1)


def main():
    a = 1664525
    b = 1013904223
    m = 2**32
    
    x0 = int(input('x0: '))
    n = int(input('кол-во чисел: '))
    
    print(f'\nсгенерированные числа:')
    print(a, b, m)
    lcg_recursive(x0, a, b, m, n)

main()
```
